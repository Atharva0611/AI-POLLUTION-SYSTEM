```python
import os
import pandas as pd


# ============================================================
# 1. PROJECT PATHS
# ============================================================

# Current working directory
BASE_DIR = os.getcwd()

# If notebook is inside the "notebooks" folder,
# move one level up to the project root
if os.path.basename(BASE_DIR).lower() == "notebooks":
    BASE_DIR = os.path.dirname(BASE_DIR)

RAW_DIR = os.path.join(
    BASE_DIR,
    "data",
    "raw",
    "cpcb"
)

PROCESSED_DIR = os.path.join(
    BASE_DIR,
    "data",
    "processed"
)

os.makedirs(PROCESSED_DIR, exist_ok=True)


# ============================================================
# 2. CPCB STATION FILES
# ============================================================

station_files = {
    "Alandi": "alandi.csv",
    "Bhosari": "bhosari.csv",
    "Bhumkar Nagar": "bhumkar_nagar.csv",
    "Dhankawadi": "dhankawadi.csv",
    "Hadapsar": "hadapsar.csv",
    "Karve Road": "karve_road.csv",
    "Katraj Dairy": "katraj_dairy.csv",
    "Mhada Colony": "mhada_colony.csv",
    "Panchawati_Pashan": "panchawati_pashan.csv",
    "Revenue Colony-Shivajinagar": "revenue_colony_shivajinagar.csv",
    "SPPU": "sppu.csv",
    "Transport Nagar-Nigdi": "transport_nagar_nigdi.csv"
}


# ============================================================
# 3. CHECK FOLDER
# ============================================================

print("=" * 70)
print("PUNE CPCB DATA COLLECTION")
print("=" * 70)

print("\nProject folder:")
print(BASE_DIR)

print("\nCPCB folder:")
print(RAW_DIR)


# ============================================================
# 4. CHECK ALL FILES
# ============================================================

print("\n" + "=" * 70)
print("CHECKING STATION FILES")
print("=" * 70)

available_files = []

for station_name, filename in station_files.items():

    file_path = os.path.join(
        RAW_DIR,
        filename
    )

    if os.path.exists(file_path):

        print(f"✅ {station_name} → {filename}")

        available_files.append(
            (station_name, file_path)
        )

    else:

        print(f"❌ {station_name} → {filename} NOT FOUND")


# ============================================================
# 5. LOAD ALL STATIONS
# ============================================================

print("\n" + "=" * 70)
print("LOADING DATA")
print("=" * 70)

all_data = []

for station_name, file_path in available_files:

    try:

        data = pd.read_csv(file_path)

        # Add station name
        data["station_name"] = station_name

        all_data.append(data)

        print(
            f"✅ {station_name}: "
            f"{len(data):,} rows"
        )

    except Exception as e:

        print(
            f"❌ Error loading "
            f"{station_name}: {e}"
        )


# ============================================================
# 6. COMBINE ALL STATIONS
# ============================================================

if len(all_data) == 0:

    raise FileNotFoundError(
        "\n❌ No CSV files found.\n"
        "Put your CPCB files inside:\n"
        + RAW_DIR
    )


df = pd.concat(
    all_data,
    ignore_index=True,
    sort=False
)


# ============================================================
# 7. DISPLAY INFORMATION
# ============================================================

print("\n" + "=" * 70)
print("COMBINED DATASET")
print("=" * 70)

print("\nTotal rows:", f"{len(df):,}")
print("Total columns:", len(df.columns))
print(
    "Total stations:",
    df["station_name"].nunique()
)


# ============================================================
# 8. STATION-WISE SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("STATION-WISE ROW COUNT")
print("=" * 70)

station_summary = (
    df["station_name"]
    .value_counts()
)

print(station_summary)


# ============================================================
# 9. SHOW FIRST 5 ROWS
# ============================================================

print("\nFirst 5 rows:")

display(df.head())


# ============================================================
# 10. SAVE COMBINED DATASET
# ============================================================

output_file = os.path.join(
    PROCESSED_DIR,
    "pune_cpcb_combined_raw.csv.gz"
)

df.to_csv(
    output_file,
    index=False
)


# ============================================================
# 11. FINAL MESSAGE
# ============================================================

print("\n" + "=" * 70)
print("✅ DATA COLLECTION COMPLETED")
print("=" * 70)

print("\nOutput saved at:")
print(output_file)

print("\nDataset shape:")
print(df.shape)
```

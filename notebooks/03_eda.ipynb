```python


import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt




BASE_DIR = os.getcwd()

if os.path.basename(BASE_DIR).lower() == "notebooks":
    BASE_DIR = os.path.dirname(BASE_DIR)

INPUT_FILE = os.path.join(
    BASE_DIR,
    "data",
    "processed",
    "pune_cpcb_cleaned_hourly.csv"
)

print("=" * 70)
print("PUNE CPCB EXPLORATORY DATA ANALYSIS")
print("=" * 70)

print("\nLoading file:")
print(INPUT_FILE)




df = pd.read_csv(INPUT_FILE)

df["timestamp"] = pd.to_datetime(
    df["timestamp"],
    errors="coerce"
)

print("\n✅ Dataset loaded successfully")
print("Shape:", df.shape)




print("\n" + "=" * 70)
print("1. BASIC INFORMATION")
print("=" * 70)

print("Total rows:", f"{len(df):,}")
print("Total columns:", len(df.columns))
print("Total stations:", df["station_name"].nunique())

print("\nStations:")

for station in sorted(df["station_name"].dropna().unique()):
    print("✅", station)


# ============================================================
# 4. COLUMN INFORMATION
# ============================================================

print("\n" + "=" * 70)
print("2. COLUMNS")
print("=" * 70)

print(df.columns.tolist())


# ============================================================
# 5. DATA TYPES
# ============================================================

print("\n" + "=" * 70)
print("3. DATA TYPES")
print("=" * 70)

print(df.dtypes)


# ============================================================
# 6. MISSING VALUES
# ============================================================

print("\n" + "=" * 70)
print("4. MISSING VALUES")
print("=" * 70)

missing = (
    df.isnull()
    .mean()
    .mul(100)
    .sort_values(ascending=False)
)

print(missing)


# ============================================================
# 7. STATION-WISE RECORD COUNT
# ============================================================

print("\n" + "=" * 70)
print("5. STATION-WISE RECORD COUNT")
print("=" * 70)

station_count = (
    df["station_name"]
    .value_counts()
    .sort_values(ascending=False)
)

print(station_count)


# ============================================================
# 8. STATION-WISE BAR CHART
# ============================================================

plt.figure(figsize=(12, 6))

station_count.plot(kind="bar")

plt.title("Number of Hourly Observations by Station")
plt.xlabel("Station")
plt.ylabel("Number of Observations")
plt.xticks(rotation=60, ha="right")

plt.tight_layout()
plt.show()


# ============================================================
# 9. DATE RANGE
# ============================================================

print("\n" + "=" * 70)
print("6. DATE RANGE")
print("=" * 70)

print("Overall start:", df["timestamp"].min())
print("Overall end:", df["timestamp"].max())

date_range = (
    df.groupby("station_name")["timestamp"]
    .agg(["min", "max"])
)

display(date_range)


# ============================================================
# 10. NUMERICAL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("7. STATISTICAL SUMMARY")
print("=" * 70)

numeric_columns = [
    "pm25",
    "pm10",
    "no2",
    "so2",
    "co",
    "ozone",
    "temperature",
    "humidity",
    "wind_speed",
    "wind_direction",
    "rainfall"
]

numeric_columns = [
    col for col in numeric_columns
    if col in df.columns
]

display(
    df[numeric_columns].describe().T
)


# ============================================================
# 11. PM2.5 DISTRIBUTION
# ============================================================

print("\n" + "=" * 70)
print("8. PM2.5 DISTRIBUTION")
print("=" * 70)

print("Mean:", round(df["pm25"].mean(), 2))
print("Median:", round(df["pm25"].median(), 2))
print("Minimum:", round(df["pm25"].min(), 2))
print("Maximum:", round(df["pm25"].max(), 2))


plt.figure(figsize=(10, 6))

plt.hist(
    df["pm25"].dropna(),
    bins=100
)

plt.title("PM2.5 Distribution")
plt.xlabel("PM2.5")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()


# ============================================================
# 12. PM2.5 BOXPLOT
# ============================================================

plt.figure(figsize=(8, 6))

plt.boxplot(
    df["pm25"].dropna()
)

plt.title("PM2.5 Boxplot")
plt.ylabel("PM2.5")

plt.tight_layout()
plt.show()


# ============================================================
# 13. PM2.5 BY STATION
# ============================================================

print("\n" + "=" * 70)
print("9. AVERAGE PM2.5 BY STATION")
print("=" * 70)

station_pm25 = (
    df.groupby("station_name")["pm25"]
    .agg([
        "count",
        "mean",
        "median",
        "min",
        "max",
        "std"
    ])
    .sort_values("mean", ascending=False)
)

display(station_pm25)


plt.figure(figsize=(12, 6))

station_pm25["mean"].plot(kind="bar")

plt.title("Average PM2.5 by Pune CPCB Station")
plt.xlabel("Station")
plt.ylabel("Average PM2.5")
plt.xticks(rotation=60, ha="right")

plt.tight_layout()
plt.show()


# ============================================================
# 14. POLLUTANT AVERAGES
# ============================================================

print("\n" + "=" * 70)
print("10. AVERAGE POLLUTANT VALUES")
print("=" * 70)

pollutants = [
    "pm25",
    "pm10",
    "no2",
    "so2",
    "co",
    "ozone"
]

pollutants = [
    col for col in pollutants
    if col in df.columns
]

pollutant_mean = (
    df[pollutants]
    .mean()
    .sort_values(ascending=False)
)

print(pollutant_mean)


plt.figure(figsize=(10, 6))

pollutant_mean.plot(kind="bar")

plt.title("Average Pollutant Concentration")
plt.xlabel("Pollutant")
plt.ylabel("Average Value")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()


# ============================================================
# 15. CORRELATION
# ============================================================

print("\n" + "=" * 70)
print("11. CORRELATION MATRIX")
print("=" * 70)

correlation_columns = [
    "pm25",
    "pm10",
    "no2",
    "so2",
    "co",
    "ozone",
    "temperature",
    "humidity",
    "wind_speed",
    "wind_direction",
    "rainfall"
]

correlation_columns = [
    col for col in correlation_columns
    if col in df.columns
]

corr = df[correlation_columns].corr()

display(corr.round(2))


# ============================================================
# 16. CORRELATION HEATMAP
# ============================================================

plt.figure(figsize=(12, 9))

plt.imshow(
    corr,
    aspect="auto"
)

plt.colorbar()

plt.xticks(
    range(len(corr.columns)),
    corr.columns,
    rotation=60,
    ha="right"
)

plt.yticks(
    range(len(corr.columns)),
    corr.columns
)

plt.title("Correlation Matrix")

plt.tight_layout()
plt.show()


# ============================================================
# 17. TIME FEATURES
# ============================================================

df["hour"] = df["timestamp"].dt.hour
df["day_of_week"] = df["timestamp"].dt.dayofweek
df["month"] = df["timestamp"].dt.month
df["date"] = df["timestamp"].dt.date


# ============================================================
# 18. PM2.5 BY HOUR
# ============================================================

print("\n" + "=" * 70)
print("12. PM2.5 BY HOUR")
print("=" * 70)

hourly_pm25 = (
    df.groupby("hour")["pm25"]
    .mean()
)

print(hourly_pm25)


plt.figure(figsize=(10, 6))

plt.plot(
    hourly_pm25.index,
    hourly_pm25.values,
    marker="o"
)

plt.title("Average PM2.5 by Hour of Day")
plt.xlabel("Hour")
plt.ylabel("Average PM2.5")

plt.xticks(range(24))
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()


# ============================================================
# 19. PM2.5 BY DAY OF WEEK
# ============================================================

print("\n" + "=" * 70)
print("13. PM2.5 BY DAY OF WEEK")
print("=" * 70)

day_names = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

weekly_pm25 = (
    df.groupby("day_of_week")["pm25"]
    .mean()
)

weekly_pm25.index = [
    day_names[i]
    for i in weekly_pm25.index
]

print(weekly_pm25)


plt.figure(figsize=(10, 6))

weekly_pm25.plot(kind="bar")

plt.title("Average PM2.5 by Day of Week")
plt.xlabel("Day")
plt.ylabel("Average PM2.5")
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()


# ============================================================
# 20. DAILY PM2.5 TREND
# ============================================================

print("\n" + "=" * 70)
print("14. DAILY PM2.5 TREND")
print("=" * 70)

daily_pm25 = (
    df.set_index("timestamp")["pm25"]
    .resample("D")
    .mean()
)

plt.figure(figsize=(14, 6))

plt.plot(
    daily_pm25.index,
    daily_pm25.values
)

plt.title("Daily Average PM2.5 — Pune")
plt.xlabel("Date")
plt.ylabel("Average PM2.5")

plt.tight_layout()
plt.show()


# ============================================================
# 21. MONTHLY PM2.5 TREND
# ============================================================

print("\n" + "=" * 70)
print("15. MONTHLY PM2.5 TREND")
print("=" * 70)

monthly_pm25 = (
    df.set_index("timestamp")["pm25"]
    .resample("MS")
    .mean()
)

plt.figure(figsize=(14, 6))

plt.plot(
    monthly_pm25.index,
    monthly_pm25.values,
    marker="o"
)

plt.title("Monthly Average PM2.5 — Pune")
plt.xlabel("Month")
plt.ylabel("Average PM2.5")

plt.tight_layout()
plt.show()


# ============================================================
# 22. ALL STATIONS DAILY PM2.5
# ============================================================

print("\n" + "=" * 70)
print("16. DAILY PM2.5 TREND — ALL STATIONS")
print("=" * 70)

plt.figure(figsize=(14, 8))

for station in sorted(df["station_name"].unique()):

    station_data = (
        df[df["station_name"] == station]
        .set_index("timestamp")["pm25"]
        .resample("D")
        .mean()
    )

    plt.plot(
        station_data.index,
        station_data.values,
        label=station
    )

plt.title("Daily PM2.5 Trend — All Pune Stations")
plt.xlabel("Date")
plt.ylabel("PM2.5")

plt.legend(
    bbox_to_anchor=(1.02, 1),
    loc="upper left"
)

plt.tight_layout()
plt.show()


# ============================================================
# 23. TOP 20 HIGHEST PM2.5
# ============================================================

print("\n" + "=" * 70)
print("17. TOP 20 HIGHEST PM2.5 OBSERVATIONS")
print("=" * 70)

top_pm25 = (
    df[
        [
            "station_name",
            "timestamp",
            "pm25"
        ]
    ]
    .sort_values(
        "pm25",
        ascending=False
    )
    .head(20)
)

display(top_pm25)



print("\n" + "=" * 70)
print("FINAL EDA SUMMARY")
print("=" * 70)

print("Total observations :", f"{len(df):,}")
print("Total stations     :", df["station_name"].nunique())
print("Start date         :", df["timestamp"].min())
print("End date           :", df["timestamp"].max())

print("\nPM2.5 statistics:")
print("Mean               :", round(df["pm25"].mean(), 2))
print("Median             :", round(df["pm25"].median(), 2))
print("Maximum            :", round(df["pm25"].max(), 2))

highest_station = station_pm25["mean"].idxmax()
highest_value = station_pm25["mean"].max()

print("\nHighest average PM2.5 station:")
print(highest_station, "→", round(highest_value, 2))

print("\n✅ EDA COMPLETED FOR ALL STATIONS")
```
